## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

## 2. Load the Analysis Dataset

The Week 1 cleaned dataset is loaded for exploratory analysis.

If `superstore_cleaned.csv` is not in the same folder as this notebook, change the path below.

```python
df = pd.read_csv("superstore_cleaned.csv")

# Convert dates again after reading the CSV
df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])

print("Dataset loaded successfully.")
print("Shape:", df.shape)
df.head()
```

## 3. Initial Dataset Overview

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## 4. Descriptive Statistics

Descriptive statistics are generated for the main numerical variables.

```python
numeric_columns = ["sales", "quantity", "discount", "profit"]

descriptive_stats = df[numeric_columns].describe().T
descriptive_stats
```

## 5. Key Overall Metrics

```python
total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
total_quantity = df["quantity"].sum()
profit_margin = (total_profit / total_sales) * 100

summary_metrics = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Quantity",
        "Overall Profit Margin",
        "Unique Orders",
        "Unique Customers",
        "Unique Products"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_quantity,
        profit_margin,
        df["order_id"].nunique(),
        df["customer_id"].nunique(),
        df["product_id"].nunique()
    ]
})

summary_metrics
```

## 6. Categorical Variable Analysis

The main categorical variables are examined using counts.

```python
categorical_columns = ["category", "sub_category", "segment", "region", "ship_mode"]

for column in categorical_columns:
    print(f"\n--- {column.upper()} ---")
    print(df[column].value_counts())
```

## 7. Category Performance

Sales, profit and profit margin are compared across product categories.

```python
category_analysis = (
    df.groupby("category")[["sales", "profit"]]
      .sum()
      .sort_values("sales", ascending=False)
)

category_analysis["profit_margin"] = (
    category_analysis["profit"] / category_analysis["sales"]
) * 100

category_analysis
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

category_analysis["sales"].plot(kind="bar", ax=ax)
ax.set_title("Sales by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

category_analysis["profit"].plot(kind="bar", ax=ax)
ax.set_title("Profit by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

category_analysis["profit_margin"].plot(kind="bar", ax=ax)
ax.set_title("Profit Margin by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Profit Margin (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. Sub-Category Performance

Sub-categories are compared to identify areas contributing strongly to sales or profit and areas requiring further investigation.

```python
subcategory_analysis = (
    df.groupby("sub_category")[["sales", "profit"]]
      .sum()
      .sort_values("profit")
)

subcategory_analysis["profit_margin"] = (
    subcategory_analysis["profit"] / subcategory_analysis["sales"]
) * 100

subcategory_analysis
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

subcategory_analysis["profit"].plot(kind="bar", ax=ax)
ax.set_title("Profit by Sub-Category")
ax.set_xlabel("Sub-Category")
ax.set_ylabel("Profit")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 9. Discount Analysis

Discount levels are compared with profit to investigate whether discounting is associated with lower profitability.

```python
discount_analysis = (
    df.groupby("discount")[["sales", "profit"]]
      .sum()
      .sort_index()
)

discount_analysis["profit_margin"] = (
    discount_analysis["profit"] / discount_analysis["sales"]
) * 100

discount_analysis
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

discount_analysis["profit"].plot(kind="bar", ax=ax)
ax.set_title("Total Profit by Discount Level")
ax.set_xlabel("Discount")
ax.set_ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

discount_analysis["profit_margin"].plot(kind="line", marker="o", ax=ax)
ax.set_title("Profit Margin by Discount Level")
ax.set_xlabel("Discount")
ax.set_ylabel("Profit Margin (%)")
plt.tight_layout()
plt.show()
```

### Discount and profit interpretation

The discount analysis should be interpreted carefully. A relationship between discount and profit does not by itself prove that discounting caused the change in profit. Product mix, region, order size and other factors may also contribute.

## 10. Relationship Between Discount and Profit

The average profit per transaction is also examined at each discount level.

```python
discount_mean_profit = df.groupby("discount")["profit"].mean().sort_index()
discount_mean_profit
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

discount_mean_profit.plot(kind="line", marker="o", ax=ax)
ax.set_title("Average Profit per Record by Discount Level")
ax.set_xlabel("Discount")
ax.set_ylabel("Average Profit")
plt.tight_layout()
plt.show()
```

## 11. Correlation Analysis

Correlation is calculated for the main numerical variables.

A correlation coefficient closer to +1 indicates a strong positive linear relationship, while a coefficient closer to -1 indicates a strong negative linear relationship. A value near 0 indicates a weak linear relationship.

Correlation does not establish causation.

```python
correlation_matrix = df[["sales", "quantity", "discount", "profit"]].corr()
correlation_matrix
```

```python
fig, ax = plt.subplots(figsize=(7, 6))

image = ax.imshow(correlation_matrix, interpolation="nearest")
ax.set_xticks(range(len(correlation_matrix.columns)))
ax.set_yticks(range(len(correlation_matrix.columns)))
ax.set_xticklabels(correlation_matrix.columns, rotation=45, ha="right")
ax.set_yticklabels(correlation_matrix.columns)
ax.set_title("Correlation Matrix: Numerical Variables")

for i in range(len(correlation_matrix.columns)):
    for j in range(len(correlation_matrix.columns)):
        ax.text(j, i, f"{correlation_matrix.iloc[i, j]:.2f}",
                ha="center", va="center")

fig.colorbar(image, ax=ax)
plt.tight_layout()
plt.show()
```

## 12. Sales vs Profit Relationship

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(df["sales"], df["profit"], alpha=0.5)
ax.axhline(0, linewidth=1)
ax.set_title("Sales vs Profit")
ax.set_xlabel("Sales")
ax.set_ylabel("Profit")
plt.tight_layout()
plt.show()
```

## 13. Time-Based Analysis

```python
fig, ax = plt.subplots(figsize=(12, 5))

monthly_analysis["profit"].plot(ax=ax)
ax.axhline(0, linewidth=1)
ax.set_title("Monthly Profit Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Profit")
plt.tight_layout()
plt.show()
```

## 14. Regional Performance

Regional sales and profit are compared to identify geographic differences.

```python
region_analysis = (
    df.groupby("region")[["sales", "profit"]]
      .sum()
      .sort_values("profit")
)

region_analysis["profit_margin"] = (
    region_analysis["profit"] / region_analysis["sales"]
) * 100

region_analysis
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

region_analysis["profit"].plot(kind="bar", ax=ax)
ax.set_title("Profit by Region")
ax.set_xlabel("Region")
ax.set_ylabel("Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 15. Investigating the Furniture / Tables Pattern

The earlier analysis identified Tables as an area requiring closer investigation. This section drills down from category → sub-category → product → region → discount.

```python
furniture = df[df["category"] == "Furniture"]

furniture_subcategory = (
    furniture.groupby("sub_category")[["sales", "profit"]]
             .sum()
             .sort_values("profit")
)

furniture_subcategory
```

```python
tables = df[df["sub_category"] == "Tables"]

table_analysis = (
    tables.groupby("discount")[["sales", "profit"]]
          .sum()
          .sort_index()
)

table_analysis["profit_margin"] = (
    table_analysis["profit"] / table_analysis["sales"]
) * 100

table_analysis
```

```python
table_products = (
    tables.groupby("product_name")[["sales", "profit", "discount"]]
          .agg({
              "sales": "sum",
              "profit": "sum",
              "discount": "mean"
          })
          .sort_values("profit")
)

table_products.head(10)
```

```python
table_region = (
    tables.groupby("region")[["sales", "profit"]]
          .sum()
          .sort_values("profit")
)

table_region["profit_margin"] = (
    table_region["profit"] / table_region["sales"]
) * 100

table_region
```

## 16. East Region — Tables Drill-Down

The East region is examined further because the earlier analysis showed a negative result for Tables in that region.

```python
east_tables = tables[tables["region"] == "East"]

east_product_analysis = (
    east_tables.groupby("product_name")[["sales", "profit", "discount"]]
               .agg({
                   "sales": "sum",
                   "profit": "sum",
                   "discount": "mean"
               })
               .sort_values("profit")
)

east_product_analysis.head(10)
```

```python
east_discount_analysis = (
    east_tables.groupby("discount")[["sales", "profit"]]
               .sum()
               .sort_index()
)

east_discount_analysis["profit_margin"] = (
    east_discount_analysis["profit"] /
    east_discount_analysis["sales"]
) * 100

east_discount_analysis
```

```python
east_state_analysis = (
    east_tables.groupby("state")[["sales", "profit"]]
               .sum()
               .sort_values("profit")
)

east_state_analysis["profit_margin"] = (
    east_state_analysis["profit"] /
    east_state_analysis["sales"]
) * 100

east_state_analysis
```

## 17. High-Loss States at 40% Discount

New York, Ohio and Pennsylvania were identified in the earlier analysis as important areas for further investigation. This section examines their 40% discount performance.

```python
top_loss_states = east_tables[
    east_tables["state"].isin(["New York", "Ohio", "Pennsylvania"])
]

state_discount_analysis = (
    top_loss_states.groupby(["state", "discount"])[["sales", "profit"]]
                   .sum()
)

state_discount_analysis["profit_margin"] = (
    state_discount_analysis["profit"] /
    state_discount_analysis["sales"]
) * 100

state_discount_analysis
```

```python
state_discount_counts = (
    top_loss_states.groupby(["state", "discount"])
                   .size()
                   .reset_index(name="records")
)

state_discount_counts
```

```python
state_order_counts = (
    top_loss_states.groupby(["state", "discount"])["order_id"]
                   .nunique()
                   .reset_index(name="unique_orders")
)

state_order_counts
```

## 18. Top Loss-Making Products

The products with the lowest total profit are listed for further investigation.

```python
top_loss_products = (
    df.groupby("product_name")[["sales", "profit"]]
      .sum()
      .sort_values("profit")
      .head(10)
)

top_loss_products
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

top_loss_products["profit"].sort_values().plot(kind="barh", ax=ax)
ax.axvline(0, linewidth=1)
ax.set_title("Top 10 Loss-Making Products")
ax.set_xlabel("Total Profit")
ax.set_ylabel("Product")
plt.tight_layout()
plt.show()
```

## 19. Five+ Data-Driven Insights

The following insights are supported by the analysis already performed. The exact values shown here are based on the results obtained during the project and can be refreshed by running the notebook.

### Insight 1 — Overall profitability

The dataset contains total sales of approximately  **$2.30  million** and total profit of approximately **$286.40 thousand**, giving an overall profit margin of approximately **12.47%**.

### Insight 2 — Technology leads sales and profit

Technology generated approximately **$836.15 thousand  in sales** and **$145.45 thousand in profit**, while also recording a profit margin of approximately **17.40%**.

### Insight 3 — Furniture has a much lower profit margin

Furniture generated approximately **$742.00  thousand  in sales**, but only approximately **$18.45 thousand in profit**, resulting in a profit margin of approximately **2.49%**. This indicates that high sales do not necessarily translate into high profitability.

### Insight 4 — Tables are a significant loss area

Within Furniture, the Tables sub-category generated approximately **$206.97 thousand in sales** but approximately **-$17.73 thousand in profit**. This makes Tables an important area for further investigation.

### Insight 5 — High discount levels require investigation

The analysis showed negative total profit at some higher discount levels. This indicates that certain discount levels are associated with loss-making transactions and should be investigated together with product, region and order characteristics.

### Insight 6 — Losses are concentrated in specific locations for the investigated Tables segment

For the East-region Tables analysis, **New York,  Ohio  and  Pennsylvania**  were identified as important loss-making states. At a **40% discount**, the previously calculated results were approximately:

- Pennsylvania: **- $2,588.75  profit**
- New York:   **- $4,535.64   profit**
- Ohio:   **- $2,715.33   profit**

These results suggest that discounting should be investigated at a more detailed product-and-location level rather than looking only at overall averages.

### Insight 7 — Individual products can materially affect category performance

The product-level analysis identified individual loss-making products within Tables, including products with losses exceeding $1,000. This demonstrates why drilling down from category to sub-category and then product level is useful in business analysis.

> **Important:** These findings describe patterns in the dataset. They do not by themselves establish that discounting caused the losses. Further analysis would be required to test causation.
